```python
def needs_charging(devices):
    result = []
    for dev in devices:
        name = dev[0]
        battery = dev[1]
        if battery < 20:
            result.append(name)
    return result
```

```python
Test.assert_equals(needs_charging([["iPhone", 15], ["MacBook", 80], ["iPad", 19]]), ["iPhone", "iPad"])

Test.assert_equals(needs_charging([["iPhone", 20], ["iPad", 0]]), ["iPad"])

Test.assert_equals(needs_charging([]), [])

Test.assert_equals(needs_charging([["AirPods", 20], ["Apple Watch", 2], ["iPad", 21]]), ["Apple Watch"])
```

```python
def find_highest(lst):
    if len(lst) == 1:
        return lst[0]
    max_rest = find_highest(lst[1:])
    if lst[0] > max_rest:
        return lst[0]
    else:
        return max_rest
```

```python
Test.assert_equals(find_highest([-1, 3, 5, 6, 99, 12, 2]), 99)

Test.assert_equals(find_highest([0, 12, 4, 87]), 87)

Test.assert_equals(find_highest([8]), 8)

Test.assert_equals(find_highest([-10,-20,-1]), -1)
```

```python
def snakefill(n):
    total_cells = n * n
    snake_len = 1
    food = 0
    while snake_len * 2 < total_cells:
        snake_len *= 2
        food += 1
    return food
```

```python
Test.assert_equals(snakefill(3), 3)

Test.assert_equals(snakefill(6), 5)

Test.assert_equals(snakefill(24), 9)

Test.assert_equals(snakefill(1), 0)
```

```python
def atbash(txt):
    result = ""
    for char in txt:
        if char.islower():
            result += chr(219-ord(char)) # 'a' - 97, 'z' - 122
        elif char.isupper():
            result += chr(155 - ord(char)) # 'A' - 65, 'Z' - 90
        else:
            result += char
    return result
```

```python
Test.assert_equals(atbash("apple"), "zkkov")

Test.assert_equals(atbash("Hello world!"), "Svool dliow!")

Test.assert_equals(atbash("Christmas is the 25th of December"), "Xsirhgnzh rh gsv 25gs lu Wvxvnyvi")

Test.assert_equals(atbash("AbC"), "ZyX")
```

```python
def longest_substring(digits):
    if not digits:
        return ""
    max_substr = digits[0]
    curr_substr = digits[0]

    for i in range(1,len(digits)):
        curr_digit = int(digits[i])
        prev_digit = int(digits[i-1])
        if (curr_digit % 2 != prev_digit % 2):
            curr_substr += digits[i]
        else:
            if len(curr_substr) > len(max_substr):
                max_substr = curr_substr
            curr_substr = digits[i]
        if len(curr_substr) > len(max_substr):
                max_substr = curr_substr
    return max_substr
```

```python
Test.assert_equals(longest_substring("225424272163254474441338664823"), "272163254")

Test.assert_equals(longest_substring("594127169973391692147228678476"), "16921472")

Test.assert_equals(longest_substring("721449827599186159274227324466"), "7214")

Test.assert_equals(longest_substring("123456789"), "123456789")
```
